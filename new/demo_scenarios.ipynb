# NurseBot — Demo Scenarios
Pre-built scenarios that run without a microphone. Run `nursebot.ipynb` first, then run cells here.

## Scenario 1 — Doctor adds tasks, robot works through queue

```python
def go(text, role="doctor", pause=1.5):
    resp = nurse.process(text, role)
    audio = tts.synth(resp)
    display(Audio(audio, rate=tts.sr, autoplay=True))  # plays in Jupyter output
    if SD_AVAILABLE:
        sd.play(audio, samplerate=tts.sr)
        sd.wait()
    time.sleep(pause)
```

## Scenario 2 — Emergency interrupt mid-task

```python
go("check blood pressure room 302")
time.sleep(3)

# This arrives while the robot is working on room 302
go("urgent — patient in room 301 is having chest pains, immediate assessment needed")
time.sleep(5)
print("Robot should pause room 302 and rush to room 301")
```

## Scenario 3 — Patient emergency request

```python
go("check blood pressure room 303")
time.sleep(2)

# Patient triggers emergency
go("I'm in severe pain, I need help right now", role="patient")
time.sleep(2)
print("Robot should interrupt and respond to patient")
```

## Scenario 4 — Permission checks

```python
# Add a task as doctor
go("change IV for patient in room 306")
time.sleep(1)

# Patient tries to cancel a doctor task — should be blocked
go("cancel the IV change", role="patient")
time.sleep(1)

# Doctor cancels it — should work
go("cancel the IV change", role="doctor")
```

## Scenario 5 — Full workflow

```python
tasks = [
    ("check blood pressure room 302", "doctor"),
    ("administer medication room 305", "doctor"),
    ("take temperature room 303", "doctor"),
]

for text, role in tasks:
    go(text, role, pause=1.0)

time.sleep(2)
go("what tasks are currently active", "doctor")
time.sleep(20)
print("\nFinal state:")
print(tm.status_report())
```
