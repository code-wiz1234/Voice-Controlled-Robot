```python
import os,json,uuid,time,threading,io,re,socket
from datetime import datetime
from dataclasses import dataclass,field
from typing import Optional,Dict,List,Callable
from enum import Enum,IntEnum
import numpy as np
import sounddevice as sd
import soundfile as sf
import scipy.signal
import requests
from faster_whisper import WhisperModel
from kokoro import KPipeline
from IPython.display import Audio,display

STT_SAMPLE_RATE=16_000;TTS_SAMPLE_RATE=24_000
WHISPER_MODEL="base.en";KOKORO_VOICE="af_heart"
OLLAMA_MODEL="llama3.2";OLLAMA_URL="http://localhost:11434/api/generate"
FLASK_PORT=7861;GRADIO_PORT=7860
ROLE_PERMS={"doctor":{"can_interrupt":True,"can_delete_any":True,"can_set_urgent":True},
            "patient":{"can_interrupt":True,"can_delete_any":False,"can_set_urgent":False}}
print("✅ Config loaded.")
```

```python
class Priority(IntEnum):
    LOW=1;NORMAL=2;URGENT=3
class TaskStatus(Enum):
    QUEUED="queued";RUNNING="running";INTERRUPTED="interrupted"
    COMPLETED="completed";DELETED="deleted"
_TASK_LIB={
    "blood pressure":{"dur":12,"steps":["Prepare cuff","Measure BP","Record"]},
    "check blood":   {"dur":12,"steps":["Prepare cuff","Measure BP","Record"]},
    "medication":    {"dur":18,"steps":["Verify Rx","Prepare dose","Confirm ID","Administer","Log"]},
    "medicine":      {"dur":15,"steps":["Verify Rx","Prepare dose","Administer","Log"]},
    "temperature":   {"dur": 6,"steps":["Prepare thermometer","Measure","Record"]},
    "draw blood":    {"dur":20,"steps":["Tourniquet","Draw sample","Label","Send to lab"]},
    "iv":            {"dur":10,"steps":["Prepare bag","Verify","Replace","Check flow"]},
    "dressing":      {"dur":15,"steps":["Gather supplies","Remove old","Clean wound","Apply new"]},
    "wound":         {"dur":15,"steps":["Gather supplies","Remove old","Clean wound","Apply new"]},
    "assessment":    {"dur":25,"steps":["Check consciousness","Assess pain","Vitals","Document"]},
    "vital":         {"dur": 8,"steps":["Check BP","Check pulse","Check O2","Record"]},
    "ecg":           {"dur":12,"steps":["Apply electrodes","Run ECG","Print results"]},
    "emergency":     {"dur":10,"steps":["Assess patient","Call physician","Stabilise","Document"]},
}
def _task_info(n):
    nl=n.lower()
    for k,v in _TASK_LIB.items():
        if k in nl: return v
    return {"dur":10,"steps":["Prepare","Execute","Finalise"]}
@dataclass
class Task:
    id:str=field(default_factory=lambda:uuid.uuid4().hex[:8])
    name:str="Unnamed Task";description:str="";priority:Priority=Priority.NORMAL
    created_by:str="doctor";status:TaskStatus=TaskStatus.QUEUED;progress:float=0.0
    created_at:datetime=field(default_factory=datetime.now)
    started_at:Optional[datetime]=None;completed_at:Optional[datetime]=None
    estimated_duration:float=10.0
print("✅ Models ready.")
```

```python
HOSPITAL_ROOMS={
    "medicine":{"label":"Medicine Cabinet"},"lab":{"label":"Laboratory"},
    "icu":{"label":"ICU"},"301":{"label":"Room 301"},"302":{"label":"Room 302"},
    "303":{"label":"Room 303"},"304":{"label":"Room 304"},"305":{"label":"Room 305"},
    "306":{"label":"Room 306"},"station":{"label":"Nurse Station"},
}
class RobotController:
    def __init__(self):
        self.current="station";self.target="station";self.status="idle"
        self.action="Standing by";self.task_name="";self.progress=0.0
        self.carrying=None;self._lock=threading.Lock()
    def _loc(self,name):
        t=name.lower()
        m=re.search(r"room\s*(\d+)",t)
        if m and m.group(1) in HOSPITAL_ROOMS: return m.group(1)
        if any(w in t for w in ["medicine","medication","administer"]): return "medicine"
        if any(w in t for w in ["lab","blood sample","draw blood"]): return "lab"
        if any(w in t for w in ["icu","critical","intensive"]): return "icu"
        return "station"
    def assign_task(self,task):
        with self._lock:
            loc=self._loc(task.name);self.target=loc;self.task_name=task.name
            self.carrying=("medicine" if loc=="medicine" else "sample" if loc=="lab" else "equipment")
            self.status="urgent" if task.priority==Priority.URGENT else "moving"
            self.action=f"Moving to {HOSPITAL_ROOMS[loc]['label']}"
    def arrive(self):
        with self._lock:
            self.current=self.target;self.status="working"
            self.action=f"Working at {HOSPITAL_ROOMS[self.current]['label']}"
    def update_progress(self,p,step):
        with self._lock:
            self.progress=p
            self.action=f"{step} — {HOSPITAL_ROOMS.get(self.current,{}).get('label','')}"
    def complete(self):
        with self._lock:
            self.target="station";self.status="returning"
            self.action="Returning to station";self.carrying=None;self.progress=1.0
    def idle(self):
        with self._lock:
            self.current="station";self.status="idle"
            self.action="Standing by";self.task_name="";self.progress=0.0;self.carrying=None
    def get_state(self):
        with self._lock:
            return {"current":self.current,"target":self.target,
                    "status":self.status,"action":self.action,"task":self.task_name,
                    "progress":round(self.progress,3),"carrying":self.carrying}
robot=RobotController()
print("✅ RobotController ready.")
```

```python
class NurseTTS:
    def __init__(self,voice=KOKORO_VOICE,lang="a"):
        print(f"🔊 Loading Kokoro ({voice})…")
        self.pipeline=KPipeline(lang_code=lang);self.voice=voice
        self._lock=threading.Lock();print("✅ Kokoro ready.")
    def _synth(self,text):
        c=[a for _,_,a in self.pipeline(text,voice=self.voice,speed=0.93)]
        return np.concatenate(c) if c else np.zeros(100,dtype=np.float32)
    def speak(self,text,blocking=True):
        with self._lock:
            audio=self._synth(text);display(Audio(audio,rate=TTS_SAMPLE_RATE,autoplay=True))
            sd.play(audio,samplerate=TTS_SAMPLE_RATE)
            if blocking: sd.wait()
    def speak_async(self,text):
        threading.Thread(target=self.speak,args=(text,),daemon=True).start()

class NurseSTT:
    def __init__(self,model_size=WHISPER_MODEL):
        print(f"👂 Loading Faster-Whisper ({model_size})…")
        self.model=WhisperModel(model_size,device="cpu",compute_type="int8");print("✅ STT ready.")
    def transcribe(self,audio,samplerate=STT_SAMPLE_RATE):
        if audio is None or len(audio)==0: return ""
        if samplerate!=STT_SAMPLE_RATE:
            n=int(len(audio)*STT_SAMPLE_RATE/samplerate)
            audio=scipy.signal.resample(audio,n).astype(np.float32)
        audio=audio.astype(np.float32)
        if audio.max()>1.0: audio=audio/32768.0
        buf=io.BytesIO();sf.write(buf,audio,STT_SAMPLE_RATE,format="WAV");buf.seek(0)
        segs,_=self.model.transcribe(buf,beam_size=5,language="en")
        return " ".join(s.text.strip() for s in segs).strip()
print("✅ TTS + STT defined.")
```

```python
_PROMPT='''Parse to JSON only. Keys: action,task_name,task_description,priority,target_task_name,response_to_user,confidence
action: add_task|interrupt_now|delete_task|list_tasks|task_status|unknown
priority: urgent|normal|low. Doctors set urgent; patient pain/help->interrupt_now.
response_to_user: warm 1-2 sentences.'''

class SmartRuleParser:
    _U={"urgent","emergency","pain","chest","help","bleeding","severe","falling"}
    _D={"cancel","delete","remove","abort"}
    _L={"list","show","status","what","tasks","queue","running"}
    _T={"blood pressure":"Check Blood Pressure","bp":"Check Blood Pressure",
        "medication":"Administer Medication","medicine":"Administer Medication",
        "temperature":"Take Temperature","vital":"Vital Signs Check",
        "draw blood":"Draw Blood Sample","iv":"Change IV",
        "dressing":"Wound Dressing","wound":"Wound Dressing",
        "assessment":"Patient Assessment","emergency":"Emergency Response"}
    def _pri(self,t,role):
        if any(w in t for w in self._U): return "urgent"
        if any(w in t for w in {"low","whenever","later"}): return "low"
        return "normal"
    def _name(self,text):
        t=text.lower()
        for kw,name in self._T.items():
            if kw in t:
                r=re.search(r"room\s*(\d+)",t)
                return f"{name} — Room {r.group(1)}" if r else name
        res=t
        for f in ["please","can you","i need","nurse","add","check on","the","a ","urgent","emergency","now"]:
            res=res.replace(f," ")
        return re.sub(r"\s+"," ",res).strip().title()[:50] or "General Task"
    def parse(self,text,role):
        t=text.lower().strip();pri=self._pri(t,role)
        if role=="patient" and any(w in t for w in {"pain","help","emergency","chest","bleeding"}):
            return {"action":"interrupt_now","task_name":"Emergency Request","task_description":text,
                    "priority":"urgent","target_task_name":None,
                    "response_to_user":"I hear you. Emergency flagged — coming right away.","confidence":0.93}
        if any(w in t for w in self._D):
            n=self._name(text)
            return {"action":"delete_task","task_name":n,"task_description":text,"priority":"normal",
                    "target_task_name":text,"response_to_user":f"Cancelling: {n}.","confidence":0.85}
        if any(w in t for w in self._L) and "blood" not in t:
            return {"action":"list_tasks","task_name":"","task_description":"","priority":"normal",
                    "target_task_name":None,"response_to_user":"Here is the task board.","confidence":0.90}
        if role=="doctor" and pri=="urgent":
            n=self._name(text)
            return {"action":"interrupt_now","task_name":n,"task_description":text,"priority":"urgent",
                    "target_task_name":None,"response_to_user":f"Urgent — moving for: {n}.","confidence":0.88}
        n=self._name(text)
        return {"action":"add_task","task_name":n,"task_description":text,"priority":pri,
                "target_task_name":None,"response_to_user":f"Adding: {n}. Priority: {pri}.","confidence":0.80}

class IntentParser:
    def __init__(self,model=OLLAMA_MODEL):
        self.model=model;self._rule=SmartRuleParser();self.use_ai=self._check()
        print(f"🧠 Intent: {'Ollama '+model if self.use_ai else 'Smart rule parser'}")
    def _check(self):
        try:
            r=requests.get("http://localhost:11434/api/tags",timeout=2)
            return r.status_code==200 and any(self.model in m["name"] for m in r.json().get("models",[]))
        except: return False
    def parse(self,text,role):
        if not self.use_ai: return self._rule.parse(text,role)
        try:
            resp=requests.post(OLLAMA_URL,json={"model":self.model,
                "prompt":f"{_PROMPT}\n\nROLE:{role.upper()}\nCOMMAND:{text}","stream":False},timeout=12)
            raw=resp.json()["response"].strip()
            s=raw.find("{");e=raw.rfind("}")+1
            return json.loads(raw[s:e]) if s>=0 and e>s else self._rule.parse(text,role)
        except: return self._rule.parse(text,role)
print("✅ Parsers ready.")
```

```python
class TaskManager:
    def __init__(self):
        self.tasks:Dict[str,Task]={};self._queue:List[Task]=[]
        self.current_task:Optional[Task]=None;self.interrupted_stack:List[Task]=[]
        self._lock=threading.RLock();self._stop_evt=threading.Event()
        self._interrupt_evt=threading.Event();self._thread:Optional[threading.Thread]=None
        self.tts:Optional[NurseTTS]=None;self._log:List[str]=[]
    def _say(self,msg):
        ts=datetime.now().strftime("%H:%M:%S");self._log.append(f"[{ts}] {msg}")
        if len(self._log)>100: self._log.pop(0)
        print(f"  🤖 {msg}")
        if self.tts: self.tts.speak_async(msg)
    def add_task(self,name,description,priority,created_by):
        with self._lock:
            t=Task(name=name,description=description,priority=priority,
                   created_by=created_by,estimated_duration=_task_info(name)["dur"])
            self.tasks[t.id]=t;self._queue.append(t)
            self._queue.sort(key=lambda x:(-x.priority.value,x.created_at))
            self._say(f"Task added: '{name}'. Priority: {priority.name.lower()}."); return t
    def interrupt_with(self,urgent_task):
        with self._lock:
            if self.current_task and self.current_task.status==TaskStatus.RUNNING:
                self.current_task.status=TaskStatus.INTERRUPTED
                self.interrupted_stack.append(self.current_task)
                self._interrupt_evt.set()
                self._say(f"Pausing '{self.current_task.name}'.")
            else: self._say(f"Executing urgent: '{urgent_task.name}'.")
            self._queue.insert(0,urgent_task)
    def delete_task(self,identifier,role):
        with self._lock:
            target=next((t for t in self.tasks.values()
                if identifier.lower() in t.name.lower() or t.id==identifier),None)
            if not target: return False,f"No task '{identifier}'."
            if role=="patient" and target.created_by=="doctor": return False,"Cannot delete doctor tasks."
            if target.status in(TaskStatus.COMPLETED,TaskStatus.DELETED): return False,"Already done."
            if target.status==TaskStatus.RUNNING:
                target.status=TaskStatus.DELETED;self._interrupt_evt.set();msg=f"Cancelling: '{target.name}'."
            elif target.status==TaskStatus.INTERRUPTED:
                target.status=TaskStatus.DELETED
                self.interrupted_stack=[t for t in self.interrupted_stack if t.id!=target.id]
                msg=f"Deleted paused: '{target.name}'."
            else:
                target.status=TaskStatus.DELETED
                self._queue=[t for t in self._queue if t.id!=target.id];msg=f"Deleted: '{target.name}'."
            self._say(msg);return True,msg
    def status_report(self):
        with self._lock:
            parts=[]
            if self.current_task: parts.append(f"Running '{self.current_task.name}' at {int(self.current_task.progress*100)}%.")
            q=[t for t in self._queue if t.status==TaskStatus.QUEUED]
            if q: parts.append("Queued: "+", ".join(t.name for t in q[:4])+".")
            if self.interrupted_stack: parts.append("Paused: "+", ".join(t.name for t in self.interrupted_stack)+".")
            return " ".join(parts) if parts else "Queue is empty."
    def _run_task(self,task):
        robot.assign_task(task)
        with self._lock:
            task.status=TaskStatus.RUNNING;task.started_at=datetime.now()
            task.progress=0.0;self.current_task=task;self._interrupt_evt.clear()
        info=_task_info(task.name);steps=info["steps"];dur=info["dur"]
        time.sleep(min(1.8,dur*0.15))
        if not self._interrupt_evt.is_set() and task.status!=TaskStatus.DELETED: robot.arrive()
        for i,step in enumerate(steps):
            if self._interrupt_evt.is_set() or self._stop_evt.is_set(): break
            robot.update_progress((i+0.5)/len(steps),step);print(f"    ⚙️  [{task.name}] {step}…")
            for j in range(10):
                if self._interrupt_evt.is_set() or self._stop_evt.is_set(): break
                time.sleep(dur/len(steps)/10);p=(i*10+j+1)/(len(steps)*10)
                task.progress=p;robot.update_progress(p,step)
        with self._lock:
            interrupted=self._interrupt_evt.is_set();self._interrupt_evt.clear()
            if task.status!=TaskStatus.DELETED and not interrupted:
                task.status=TaskStatus.COMPLETED;task.completed_at=datetime.now();task.progress=1.0
                self._say(f"Completed: '{task.name}'.");robot.complete();time.sleep(1.5);robot.idle()
                if self.interrupted_stack:
                    resume=self.interrupted_stack.pop();resume.status=TaskStatus.QUEUED;resume.progress=0.0
                    self._queue.insert(0,resume);self._say(f"Resuming: '{resume.name}'.")
            elif task.status==TaskStatus.DELETED: robot.complete();time.sleep(0.8);robot.idle()
            self.current_task=None
    def _worker_loop(self):
        while not self._stop_evt.is_set():
            with self._lock:
                nxt=self._queue.pop(0) if self._queue and self.current_task is None else None
            if nxt: self._run_task(nxt)
            else: time.sleep(0.3)
    def start(self):
        self._stop_evt.clear()
        self._thread=threading.Thread(target=self._worker_loop,name="NurseWorker",daemon=True)
        self._thread.start();print("✅ TaskManager started.")
    def stop(self):
        self._stop_evt.set()
        if self._thread: self._thread.join(timeout=3)
print("✅ TaskManager defined.")
```

```python
class RobotNurseSystem:
    def __init__(self,tm,tts,stt,parser):
        self.tm=tm;self.tts=tts;self.stt=stt;self.parser=parser;tm.tts=tts
    def process(self,text,role):
        if not text or not text.strip(): return "No command."
        print(f"\n[{role.upper()}] ▸ \"{text}\"")
        intent=self.parser.parse(text,role);action=intent.get("action","unknown")
        response=intent.get("response_to_user","Command received.");perms=ROLE_PERMS[role]
        if action=="add_task":
            p_str=intent.get("priority","normal")
            if p_str=="urgent" and not perms["can_set_urgent"]: p_str="normal"
            p_map={"urgent":Priority.URGENT,"normal":Priority.NORMAL,"low":Priority.LOW}
            priority=p_map.get(p_str,Priority.NORMAL);name=intent.get("task_name","Unspecified")
            if priority==Priority.URGENT:
                t=Task(name=name,description=intent.get("task_description",""),priority=priority,
                       created_by=role,estimated_duration=_task_info(name)["dur"])
                self.tm.tasks[t.id]=t;self.tm.interrupt_with(t)
            else: self.tm.add_task(name,intent.get("task_description",""),priority,role)
        elif action=="interrupt_now":
            if not perms["can_interrupt"]: response="No permission to interrupt."
            else:
                name=intent.get("task_name") or "Emergency Request"
                t=Task(name=name,description=intent.get("task_description",text),priority=Priority.URGENT,
                       created_by=role,estimated_duration=_task_info(name)["dur"])
                self.tm.tasks[t.id]=t;self.tm.interrupt_with(t)
        elif action=="delete_task":
            target=intent.get("target_task_name") or intent.get("task_name","")
            ok,msg=self.tm.delete_task(target,role)
            if not ok: response=msg
        elif action in("list_tasks","task_status"): response=self.tm.status_report()
        print(f"  🤖 {response}"); return response
print("✅ RobotNurseSystem defined.")
```

```python
import gradio as gr

def get_task_board():
    parts=[]
    with tm._lock:
        if tm.current_task:
            t=tm.current_task;pct=int(t.progress*100)
            icon={"URGENT":"🔴","NORMAL":"🟡","LOW":"🟢"}.get(t.priority.name,"⚪")
            parts.append(f'<div style="background:#e8f4fd;border-left:4px solid #1565c0;border-radius:8px;padding:10px;margin-bottom:8px">'
                f'<div style="display:flex;justify-content:space-between"><b>▶ {t.name}</b>'
                f'<span style="background:#1565c0;color:#fff;font-size:10px;padding:2px 8px;border-radius:99px">RUNNING</span></div>'
                f'<div style="font-size:11px;color:#555;margin:3px 0">{icon} by {t.created_by}</div>'
                f'<div style="background:#bbdefb;height:8px;border-radius:4px;overflow:hidden">'
                f'<div style="background:#1565c0;width:{pct}%;height:100%;border-radius:4px;transition:width 0.4s"></div></div>'
                f'<div style="font-size:12px;color:#1565c0;font-weight:600;margin-top:3px">{pct}%</div></div>')
        for t in reversed(tm.interrupted_stack):
            parts.append(f'<div style="background:#fff8e1;border-left:4px solid #f57f17;border-radius:8px;padding:9px;margin-bottom:6px">'
                f'<div style="display:flex;justify-content:space-between"><b>⏸ {t.name}</b>'
                f'<span style="background:#f57f17;color:#fff;font-size:10px;padding:2px 8px;border-radius:99px">PAUSED</span></div>'
                f'<div style="font-size:11px;color:#666">{int(t.progress*100)}% saved</div></div>')
        for t in tm._queue:
            bg,bc,lbl=(("#ffebee","#c62828","URGENT") if t.priority==Priority.URGENT
                       else ("#fffde7","#f9a825","NORMAL") if t.priority==Priority.NORMAL
                       else ("#e8f5e9","#388e3c","LOW"))
            parts.append(f'<div style="background:{bg};border-left:4px solid {bc};border-radius:8px;padding:8px;margin-bottom:5px">'
                f'<div style="display:flex;justify-content:space-between">'
                f'<span style="font-size:12px;font-weight:500">{t.name}</span>'
                f'<span style="background:{bc};color:#fff;font-size:10px;padding:1px 7px;border-radius:99px">{lbl}</span></div>'
                f'<div style="font-size:11px;color:#666">by {t.created_by}</div></div>')
        done=[t for t in tm.tasks.values() if t.status==TaskStatus.COMPLETED]
        for t in done[-3:]:
            parts.append(f'<div style="opacity:0.6;background:#f1f8e9;border-left:3px solid #81c784;'
                f'border-radius:8px;padding:6px;margin-bottom:4px;font-size:11px">'
                f'&#x2705; <b>{t.name}</b> &middot; by {t.created_by}</div>')
        if not tm.current_task and not tm._queue and not tm.interrupted_stack:
            parts.append('<div style="text-align:center;color:#aaa;padding:20px;font-size:13px">&#x1F4A4; Idle</div>')
    return '<div style="font-family:-apple-system,sans-serif">'+"".join(parts)+"</div>"

def get_log():
    return "\n".join(reversed(list(tm._log)[-18:])) if tm._log else "No activity yet."

def process_text(text,role):
    if not text or not text.strip():
        return "",get_task_board(),get_log(),None,"Type a command"
    response=nurse.process(text.strip(),role.lower())
    audio=tts._synth(response)
    return "",get_task_board(),get_log(),(TTS_SAMPLE_RATE,audio),f"Sent: {text.strip()}"

def process_voice(audio_data,role):
    if audio_data is None:
        return get_task_board(),get_log(),None,"No audio — record then stop"
    try:
        orig_sr,data=audio_data
        if data.ndim>1: data=data.mean(axis=1)
        data=data.astype(np.float32)
        if data.max()>1.0: data=data/32768.0
        text=stt.transcribe(data,samplerate=orig_sr)
        if not text:
            return get_task_board(),get_log(),None,"Could not hear — speak louder"
        response=nurse.process(text,role.lower())
        audio_out=tts._synth(response)
        return get_task_board(),get_log(),(TTS_SAMPLE_RATE,audio_out),f'Heard: "{text}"'
    except Exception as e:
        return get_task_board(),get_log(),None,f"Error: {e}"

def refresh(): return get_task_board(),get_log()
def quick(cmd,role): return process_text(cmd,role)

css=".big{min-height:52px!important;font-size:14px!important;font-weight:600!important}.emg{min-height:58px!important;font-size:16px!important;font-weight:700!important}"

with gr.Blocks(title="Robot Nurse",theme=gr.themes.Soft(primary_hue="blue"),css=css) as demo:

    gr.HTML(f'<div style="text-align:center;padding:12px 0 8px;border-bottom:1px solid #e5e7eb;margin-bottom:12px">'
            f'<h1 style="font-size:22px;font-weight:700;margin:0">&#x1F3E5; Robot Nurse Simulation</h1>'
            f'<p style="color:#6b7280;font-size:12px;margin:4px 0 0">'
            f'Robot map at <a href="http://localhost:{FLASK_PORT}/" target="_blank" style="color:#3b82f6">'
            f'localhost:{FLASK_PORT}</a> (opens automatically)</p></div>')

    with gr.Row(equal_height=False):
        with gr.Column(scale=6):
            # iframe loads the Flask map page — completely independent from Gradio
            gr.HTML(f'<iframe src="http://localhost:{FLASK_PORT}/" width="100%" height="455" '
                    f'style="border:none;border-radius:12px;overflow:hidden;display:block;'
                    f'background:#1a2035"></iframe>')
            gr.HTML('<p style="font-weight:600;font-size:14px;margin:10px 0 6px">&#x1F4CB; Task queue</p>')
            task_board=gr.HTML(value=get_task_board())

        with gr.Column(scale=3,min_width=280):
            role=gr.Radio(choices=["Doctor","Patient"],value="Doctor",label="Role",
                          info="Doctor=full control &#x00B7; Patient=requests+emergency")
            gr.HTML('<hr style="margin:10px 0;border-color:#e5e7eb">')
            gr.HTML('<p style="font-weight:600;font-size:13px;margin:0 0 4px">&#x2328;&#xFE0F; Text command</p>')
            text_in=gr.Textbox(placeholder="check blood pressure room 302 &#x00B7; urgent room 301 &#x00B7; cancel medication",label="",lines=2)
            send_btn=gr.Button("Send Command",variant="primary",elem_classes=["big"])
            gr.HTML('<hr style="margin:10px 0;border-color:#e5e7eb">')
            gr.HTML('<p style="font-weight:600;font-size:13px;margin:0 0 4px">&#x1F3A4; Voice command &#x2014; auto-processes when recording stops</p>')
            voice_in=gr.Audio(sources=["microphone"],type="numpy",label="",show_label=False)
            heard=gr.Textbox(label="Status",lines=1,interactive=False,value="Record then stop to process")
            gr.HTML('<hr style="margin:10px 0;border-color:#e5e7eb">')
            gr.HTML('<p style="font-weight:600;font-size:13px;margin:0 0 6px">&#x1F50A; Nurse response</p>')
            audio_out=gr.Audio(label="",autoplay=True,interactive=False,show_label=False)
            gr.HTML('<hr style="margin:10px 0;border-color:#e5e7eb">')
            gr.HTML('<p style="font-weight:600;font-size:13px;margin:0 0 6px">Quick actions</p>')
            with gr.Row():
                qbp=gr.Button("Check BP",variant="secondary")
                qmd=gr.Button("Medication",variant="secondary")
                qtp=gr.Button("Temperature",variant="secondary")
            with gr.Row():
                qvt=gr.Button("Vitals",variant="secondary")
                qas=gr.Button("Assessment",variant="secondary")
                qst=gr.Button("Status",variant="secondary")
            qhelp=gr.Button("&#x1F198; EMERGENCY",variant="stop",elem_classes=["emg"])
            gr.HTML('<hr style="margin:10px 0;border-color:#e5e7eb">')
            log_box=gr.Textbox(value="",label="&#x1F4DC; Activity log",lines=8,interactive=False,max_lines=12)

    timer=gr.Timer(value=0.5)
    timer.tick(fn=refresh,outputs=[task_board,log_box])

    txt_outs=[text_in,task_board,log_box,audio_out,heard]
    send_btn.click(fn=process_text,inputs=[text_in,role],outputs=txt_outs)
    text_in.submit(fn=process_text,inputs=[text_in,role],outputs=txt_outs)
    voice_in.change(fn=process_voice,inputs=[voice_in,role],outputs=[task_board,log_box,audio_out,heard])

    for btn,cmd in [(qbp,"check blood pressure"),(qmd,"administer medication"),
                    (qtp,"take temperature"),(qvt,"vital signs check"),
                    (qas,"patient assessment"),(qst,"task status"),
                    (qhelp,"emergency I need help now")]:
        btn.click(fn=lambda r,c=cmd:quick(c,r),inputs=[role],outputs=txt_outs)

import webbrowser
webbrowser.open(f"http://localhost:{FLASK_PORT}/")
print(f"\n&#x1F680; Launching Gradio UI on port {GRADIO_PORT}…")
print(f"   Robot map: http://localhost:{FLASK_PORT}/  (opened automatically)")
print(f"   Controls:  http://localhost:{GRADIO_PORT}/\n")
demo.launch(inbrowser=True,server_port=GRADIO_PORT,share=False,quiet=True)
```
