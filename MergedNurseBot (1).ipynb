````python
class Priority(IntEnum):
    LOW    = 1
    NORMAL = 2
    URGENT = 3

class TaskStatus(Enum):
    QUEUED      = "queued"
    RUNNING     = "running"
    INTERRUPTED = "interrupted"
    COMPLETED   = "completed"
    DELETED     = "deleted"

@dataclass
class Task:
    id:           str        = field(default_factory=lambda: uuid.uuid4().hex[:8])
    name:         str        = "Unnamed Task"
    description:  str        = ""
    priority:     Priority   = Priority.NORMAL
    created_by:   str        = "doctor"
    status:       TaskStatus = TaskStatus.QUEUED
    progress:     float      = 0.0
    created_at:   datetime   = field(default_factory=datetime.now)
    started_at:   Optional[datetime] = None
    completed_at: Optional[datetime] = None
    estimated_duration: float = 10.0

print("✅ Data models defined")
````

````python
class RobotController:
    def __init__(self):
        self.current   = "station"
        self.target    = "station"
        self.status    = "idle"
        self.action    = "Standing by"
        self.task_name = ""
        self.progress  = 0.0
        self.carrying  = None
        self._lock     = threading.Lock()

    def _dest(self, name):
        t = name.lower()
        m = re.search(r"room\s*(\d+)", t)
        if m and m.group(1) in HOSPITAL_ROOMS:
            return m.group(1)
        if any(w in t for w in ["medicine", "medication", "administer"]):
            return "medicine"
        if any(w in t for w in ["lab", "blood sample", "draw blood"]):
            return "lab"
        if any(w in t for w in ["icu", "critical", "intensive"]):
            return "icu"
        return "station"

    def assign_task(self, task):
        with self._lock:
            loc = self._dest(task.name)
            self.target    = loc
            self.task_name = task.name
            self.carrying  = "medicine" if loc == "medicine" else "sample" if loc == "lab" else "equipment"
            self.status    = "urgent" if task.priority == Priority.URGENT else "moving"
            self.action    = f"Moving to {HOSPITAL_ROOMS[loc]['label']}"

    def arrive(self):
        with self._lock:
            self.current = self.target
            self.status  = "working"
            self.action  = f"Working at {HOSPITAL_ROOMS[self.current]['label']}"

    def update_progress(self, p, step):
        with self._lock:
            self.progress = p
            lbl = HOSPITAL_ROOMS.get(self.current, {}).get("label", "")
            self.action = f"{step} — {lbl}"

    def complete(self):
        with self._lock:
            self.target   = "station"
            self.status   = "returning"
            self.action   = "Returning to nurse station"
            self.carrying = None
            self.progress = 1.0

    def idle(self):
        with self._lock:
            self.current   = "station"
            self.status    = "idle"
            self.action    = "Standing by"
            self.task_name = ""
            self.progress  = 0.0
            self.carrying  = None

    def get_state(self):
        with self._lock:
            return {
                "current":  self.current,
                "target":   self.target,
                "status":   self.status,
                "action":   self.action,
                "task":     self.task_name,
                "progress": round(self.progress, 3),
                "carrying": self.carrying,
            }

print("✅ RobotController defined")
````

````python
# ── TTSEngine: Kokoro (RTF ~0.83 on CPU — real-time capable per benchmark) ──

class TTSEngine:
    def __init__(self):
        from kokoro import KPipeline
        print("Loading Kokoro TTS...")
        self.pipeline     = KPipeline(lang_code='a') 
        self.sr           = 24_000
        self._lock        = threading.Lock()
        self._audio_queue = []   

        
        self._cache = {}
        warmup = [
            "NurseBot online. Ready for commands.",
            "Task added.",
            "On my way.",
            "Task completed.",
            "Emergency received. Responding now.",
            "Resuming previous task.",
            "Queue is empty.",
        ]
        print("Pre-caching phrases...")
        for phrase in warmup:
            self._cache[phrase] = self._generate(phrase)
        print(f"✅ TTS ready (Kokoro) — {len(self._cache)} phrases cached")

    def _generate(self, text):
        if len(text) > 150:
            text = text[:147] + "..."
        samples = []
        for _, _, audio in self.pipeline(text, voice='af_heart'):
            samples.append(audio)
        return np.concatenate(samples).astype(np.float32)

    def synth(self, text):
        if text in self._cache:
            return self._cache[text]
        audio = self._generate(text)
        self._cache[text] = audio   # cache for next time
        return audio

    def speak(self, text, blocking=True):
        with self._lock:
            audio = self.synth(text)
            self._audio_queue.append((self.sr, audio.copy()))  # feeds Gradio bg audio
            if SD_AVAILABLE:
                sd.play(audio, samplerate=self.sr)
                if blocking:
                    sd.wait()

    def speak_async(self, text):
        threading.Thread(target=self.speak, args=(text,), daemon=True).start()

    def pop_audio(self):
        if self._audio_queue:
            return self._audio_queue.pop(0)
        return None
````

````python
# ── STTEngine: Whisper small.en (better WER than base.en per benchmark) ──

class STTEngine:
    def __init__(self, model_size=WHISPER_MODEL):
        print(f"Loading Whisper {model_size}...")
        self.model = WhisperModel(model_size, device="cpu", compute_type="int8")
        print("✅ STT ready")

    def transcribe(self, audio, samplerate=STT_SR):
        if audio is None or len(audio) == 0:
            return ""
        if samplerate != STT_SR:
            n     = int(len(audio) * STT_SR / samplerate)
            audio = scipy.signal.resample(audio, n)
        audio = audio.astype(np.float32)
        if audio.ndim > 1:
            audio = audio.mean(axis=1)
        buf = io.BytesIO()
        sf.write(buf, audio, STT_SR, format="WAV")
        buf.seek(0)
        segs, _ = self.model.transcribe(buf, beam_size=5, language="en")
        return " ".join(s.text.strip() for s in segs).strip()

print("✅ STTEngine defined")
````

````python
class RuleParser:
    _ADD   = re.compile(
        r"\b(check|take|administer|change|draw|run|do|perform|start|give|go|send|bring|move|navigate|assess|measure|replace|apply)\b", re.I
    )
    _DEL   = re.compile(r"\b(cancel|delete|remove|stop|abort)\b", re.I)
    _LIST  = re.compile(r"\b(list|show|what|status|queue|tasks|active)\b", re.I)
    _EMERG = re.compile(r"\b(urgent|emergency|immediately|pain|help|chest|critical|crash|hurry|fast)\b", re.I)

    def _priority(self, t, role):
        return "urgent" if self._EMERG.search(t) and role == "doctor" else "normal"

    def _name(self, text):
        text = re.sub(
            r"\b(please|can you|could you|I need|I want|robot|nurse|hey|ok|okay)\b",
            "", text, flags=re.I
        )
        text = re.sub(r"\s+", " ", text).strip()
        return text[:60] if text else "Unspecified Task"

    def parse(self, text, role):
        t = text.lower().strip()

        # Patient emergency — always interrupt
        if self._EMERG.search(t) and role == "patient":
            return {"action": "interrupt_now",
                    "task_name": "Patient emergency",
                    "task_description": text,
                    "priority": "urgent",
                    "response_to_user": "Emergency received. I'm on my way immediately.",
                    "confidence": 0.9}

        # Urgent doctor command
        if self._EMERG.search(t) and role == "doctor":
            return {"action": "add_task",
                    "task_name": self._name(text),
                    "task_description": text,
                    "priority": "urgent",
                    "response_to_user": f"Urgent task added: '{self._name(text)}'. Responding now.",
                    "confidence": 0.85}

        # Delete / cancel
        if self._DEL.search(t):
            return {"action": "delete_task",
                    "task_name": self._name(text),
                    "target_task_name": self._name(text),
                    "priority": "normal",
                    "response_to_user": "Understood. Cancelling that task now.",
                    "confidence": 0.8}

        # Status / list
        if self._LIST.search(t):
            return {"action": "list_tasks",
                    "task_name": "",
                    "priority": "normal",
                    "response_to_user": "Here is the current task status.",
                    "confidence": 0.85}

        # Navigation / go to commands  ← NEW
        if re.search(r"\b(go|move|navigate|head|proceed|travel)\b", t, re.I):
            return {"action": "add_task",
                    "task_name": self._name(text),
                    "task_description": text,
                    "priority": "normal",
                    "response_to_user": f"On my way. Heading to {self._name(text)}.",
                    "confidence": 0.8}

        # General action command
        if self._ADD.search(t):
            return {"action": "add_task",
                    "task_name": self._name(text),
                    "task_description": text,
                    "priority": self._priority(t, role),
                    "response_to_user": f"Got it. Adding '{self._name(text)}' to the queue.",
                    "confidence": 0.75}

        # Last resort — treat any noun phrase as a task rather than saying "clarify"
        if len(t.split()) >= 2:
            return {"action": "add_task",
                    "task_name": self._name(text),
                    "task_description": text,
                    "priority": "normal",
                    "response_to_user": f"Adding '{self._name(text)}' to my queue.",
                    "confidence": 0.5}

        return {"action": "unknown",
                "task_name": text,
                "priority": "normal",
                "response_to_user": "Could you clarify that command?",
                "confidence": 0.2}

print("✅ RuleParser defined")
````

````python

from groq import Groq

GROQ_API_KEY = "Not using API Cause GIT is blocking it "   # ← paste your new key here

class GroqIntentParser:
    def __init__(self):
        self.client    = Groq(api_key=GROQ_API_KEY)
        self._fallback = RuleParser()
        print("✅ Groq LLaMA 3.3 ready")

    def parse(self, text, role="doctor"):
        prompt = f"""Parse this hospital robot voice command. Return ONLY compact JSON, no markdown.
Keys: action, task_name, task_description, priority, target_task_name, response_to_user, confidence
action: add_task | interrupt_now | delete_task | list_tasks | task_status | unknown
priority: urgent | normal | low
Only doctors may set urgent. Patient pain/help → interrupt_now urgent.
response_to_user: 1-2 warm professional sentences the robot speaks aloud.

Role: {role}
Command: {text}"""

        try:
            response = self.client.chat.completions.create(
                model="llama-3.3-70b-versatile",
                messages=[{"role": "user", "content": prompt}],
                temperature=0.1,
                max_tokens=200,
            )
            raw = response.choices[0].message.content.strip()
            raw = raw.replace("```json", "").replace("```", "").strip()
            return json.loads(raw)
        except Exception as e:
            print(f"  ⚠️ Groq failed ({e}) — falling back to rule parser")
            return self._fallback.parse(text, role)

print("✅ GroqIntentParser defined")
````

````python
class TaskManager:
    def __init__(self, robot):
        self.robot             = robot
        self.tasks: Dict[str, Task] = {}
        self._queue: List[Task]     = []
        self.current_task: Optional[Task] = None
        self.interrupted_stack: List[Task] = []
        self._log: List[str]   = []
        self._lock             = threading.RLock()
        self._stop_evt         = threading.Event()
        self._interrupt_evt    = threading.Event()
        self._thread           = None
        self.tts               = None

    def _say(self, msg):
        ts = datetime.now().strftime("%H:%M:%S")
        self._log.append(f"[{ts}] {msg}")
        if len(self._log) > 100:
            self._log.pop(0)
        print(f"  Nurse: {msg}")
        if self.tts:
            self.tts.speak_async(msg)

    def add_task(self, name, description, priority, created_by):
        with self._lock:
            t = Task(name=name, description=description, priority=priority,
                     created_by=created_by, estimated_duration=task_info(name)["dur"])
            self.tasks[t.id] = t
            self._queue.append(t)
            self._queue.sort(key=lambda x: (-x.priority.value, x.created_at))
            self._say(f"Task added: '{name}'. Priority: {priority.name.lower()}.")
            return t

    def interrupt_with(self, urgent):
        with self._lock:
            if self.current_task and self.current_task.status == TaskStatus.RUNNING:
                self.current_task.status = TaskStatus.INTERRUPTED
                self.interrupted_stack.append(self.current_task)
                self._interrupt_evt.set()
                self._say(f"Pausing '{self.current_task.name}'. Urgent: '{urgent.name}'.")
            else:
                self._say(f"Executing urgent: '{urgent.name}'.")
            self._queue.insert(0, urgent)

    def delete_task(self, identifier, role):
        with self._lock:
            target = next(
                (t for t in self.tasks.values()
                 if identifier.lower() in t.name.lower() or t.id == identifier),
                None,
            )
            if not target:
                return False, f"No task matching '{identifier}'."
            if role == "patient" and target.created_by == "doctor":
                return False, "Patients cannot delete doctor-assigned tasks."
            if target.status in (TaskStatus.COMPLETED, TaskStatus.DELETED):
                return False, f"Task already {target.status.value}."
            if target.status == TaskStatus.RUNNING:
                target.status = TaskStatus.DELETED
                self._interrupt_evt.set()
                msg = f"Cancelling running task: '{target.name}'."
            elif target.status == TaskStatus.INTERRUPTED:
                target.status = TaskStatus.DELETED
                self.interrupted_stack = [t for t in self.interrupted_stack if t.id != target.id]
                msg = f"Deleted paused task: '{target.name}'."
            else:
                target.status = TaskStatus.DELETED
                self._queue = [t for t in self._queue if t.id != target.id]
                msg = f"Deleted queued task: '{target.name}'."
            self._say(msg)
            return True, msg

    def status_report(self):
        with self._lock:
            parts = []
            if self.current_task:
                parts.append(f"Running '{self.current_task.name}' at {int(self.current_task.progress * 100)}%.")
            q = [t for t in self._queue if t.status == TaskStatus.QUEUED]
            if q:
                parts.append("Queued: " + ", ".join(t.name for t in q[:4]) + ".")
            if self.interrupted_stack:
                parts.append("Paused: " + ", ".join(t.name for t in self.interrupted_stack) + ".")
            return " ".join(parts) if parts else "Queue is empty."

    def _run_task(self, task):
        self.robot.assign_task(task)
        with self._lock:
            task.status     = TaskStatus.RUNNING
            task.started_at = datetime.now()
            task.progress   = 0.0
            self.current_task = task
            self._interrupt_evt.clear()

        info  = task_info(task.name)
        steps = info["steps"]
        dur   = info["dur"] / TASK_SPEED   # ← slowed down by TASK_SPEED multiplier

        time.sleep(min(1.8, dur * 0.15))
        if not self._interrupt_evt.is_set() and task.status != TaskStatus.DELETED:
            self.robot.arrive()

        for i, step in enumerate(steps):
            if self._interrupt_evt.is_set() or self._stop_evt.is_set():
                break
            self.robot.update_progress((i + 0.5) / len(steps), step)
            print(f"  [{task.name}] {step}")
            for j in range(10):
                if self._interrupt_evt.is_set() or self._stop_evt.is_set():
                    break
                time.sleep(dur / len(steps) / 10)   # ← checks interrupt every slice
                p = (i * 10 + j + 1) / (len(steps) * 10)
                task.progress = p
                self.robot.update_progress(p, step)

        with self._lock:
            interrupted = self._interrupt_evt.is_set()
            self._interrupt_evt.clear()
            if task.status != TaskStatus.DELETED and not interrupted:
                task.status       = TaskStatus.COMPLETED
                task.completed_at = datetime.now()
                task.progress     = 1.0
                self._say(f"Completed: '{task.name}'.")
                self.robot.complete()
                time.sleep(1.5)
                self.robot.idle()
                if self.interrupted_stack:
                    resume = self.interrupted_stack.pop()
                    resume.status   = TaskStatus.QUEUED
                    resume.progress = 0.0
                    self._queue.insert(0, resume)
                    self._say(f"Resuming: '{resume.name}'.")
            elif task.status == TaskStatus.DELETED:
                self.robot.complete()
                time.sleep(0.8)
                self.robot.idle()
            self.current_task = None

    def _worker(self):
        while not self._stop_evt.is_set():
            with self._lock:
                nxt = self._queue.pop(0) if self._queue and self.current_task is None else None
            if nxt:
                self._run_task(nxt)
            else:
                time.sleep(0.3)

    def start(self):
        self._stop_evt.clear()
        self._thread = threading.Thread(target=self._worker, name="NurseWorker", daemon=True)
        self._thread.start()

    def stop(self):
        self._stop_evt.set()
        if self._thread:
            self._thread.join(timeout=3)

print("✅ TaskManager defined")    

````

````python
class NurseSystem:
    def __init__(self, tm, tts, stt, parser):
        self.tm     = tm
        self.tts    = tts
        self.stt    = stt
        self.parser = parser
        tm.tts = tts

    def _handle(self, text, role):
        
        intent   = self.parser.parse(text, role)
        action   = intent.get("action", "unknown")
        response = intent.get("response_to_user", "Command received.")
        perms    = ROLE_PERMS[role]

        if action == "add_task":
            p_str = intent.get("priority", "normal")
            if p_str == "urgent" and not perms["can_set_urgent"]:
                p_str = "normal"
            p_map    = {"urgent": Priority.URGENT, "normal": Priority.NORMAL, "low": Priority.LOW}
            priority = p_map.get(p_str, Priority.NORMAL)
            name     = intent.get("task_name", "Unspecified Task")
            if priority == Priority.URGENT:
                t = Task(name=name, description=intent.get("task_description", ""),
                         priority=priority, created_by=role,
                         estimated_duration=task_info(name)["dur"])
                self.tm.tasks[t.id] = t
                self.tm.interrupt_with(t)
            else:
                self.tm.add_task(name, intent.get("task_description", ""), priority, role)

        elif action == "interrupt_now":
            if not perms["can_interrupt"]:
                response = "You do not have permission to interrupt."
            else:
                name = intent.get("task_name") or "Emergency Request"
                t = Task(name=name, description=intent.get("task_description", text),
                         priority=Priority.URGENT, created_by=role,
                         estimated_duration=task_info(name)["dur"])
                self.tm.tasks[t.id] = t
                self.tm.interrupt_with(t)

        elif action == "delete_task":
            target = intent.get("target_task_name") or intent.get("task_name", "")
            ok, msg = self.tm.delete_task(target, role)
            if not ok:
                response = msg

        elif action in ("list_tasks", "task_status"):
            response = self.tm.status_report()

        print(f"  Nurse: {response}")
        self.tts.speak_async(response)
        return response

    def process(self, text, role):
        if not text or not text.strip():
            return "No command received."
        print(f"\n[{role.upper()}] {text}")
       
        threading.Thread(
            target=self._handle, args=(text, role), daemon=True
        ).start()
        return "Processing..."  

    def voice_round(self, role="doctor", duration=5):
        if not SD_AVAILABLE:
            print("No audio device available.")
            return
        print(f"🎤 Listening for {duration}s ({role})...")
        audio = sd.rec(int(duration * STT_SR), samplerate=STT_SR, channels=1, dtype="float32")
        sd.wait()
        text = self.stt.transcribe(audio.flatten(), samplerate=STT_SR)
        print(f'  Heard: "{text}"')
        if text:
            self._handle(text, role)
        else:
            print("  Nothing heard.")

print("✅ NurseSystem defined")
````

````python
import gradio as gr

def _board(tm):
    parts = []
    with tm._lock:
        if tm.current_task:
            t   = tm.current_task
            pct = int(t.progress * 100)
            icon = {"URGENT": "🔴", "NORMAL": "🟡", "LOW": "🟢"}.get(t.priority.name, "⚪")
            parts.append(
                f'<div style="background:#e8f4fd;border-left:4px solid #1565c0;border-radius:8px;padding:10px;margin-bottom:8px">'
                f'<div style="display:flex;justify-content:space-between"><b>&#9654; {t.name}</b>'
                f'<span style="background:#1565c0;color:#fff;font-size:10px;padding:2px 8px;border-radius:99px">RUNNING</span></div>'
                f'<div style="font-size:11px;color:#555;margin:3px 0">{icon} by {t.created_by}</div>'
                f'<div style="background:#bbdefb;height:8px;border-radius:4px;overflow:hidden">'
                f'<div style="background:#1565c0;width:{pct}%;height:100%;border-radius:4px;transition:width .4s"></div></div>'
                f'<div style="font-size:12px;color:#1565c0;font-weight:600;margin-top:3px">{pct}%</div></div>'
            )
        for t in reversed(tm.interrupted_stack):
            parts.append(
                f'<div style="background:#fff8e1;border-left:4px solid #f57f17;border-radius:8px;padding:9px;margin-bottom:6px">'
                f'<div style="display:flex;justify-content:space-between"><b>&#9208; {t.name}</b>'
                f'<span style="background:#f57f17;color:#fff;font-size:10px;padding:2px 8px;border-radius:99px">PAUSED</span></div>'
                f'<div style="font-size:11px;color:#666">{int(t.progress*100)}% saved</div></div>'
            )
        for t in tm._queue:
            bg, bc, lbl = (
                ("#ffebee", "#c62828", "URGENT") if t.priority == Priority.URGENT else
                ("#fffde7", "#f9a825", "NORMAL") if t.priority == Priority.NORMAL else
                ("#e8f5e9", "#388e3c", "LOW")
            )
            parts.append(
                f'<div style="background:{bg};border-left:4px solid {bc};border-radius:8px;padding:8px;margin-bottom:5px">'
                f'<div style="display:flex;justify-content:space-between">'
                f'<span style="font-size:12px;font-weight:500">{t.name}</span>'
                f'<span style="background:{bc};color:#fff;font-size:10px;padding:1px 7px;border-radius:99px">{lbl}</span></div>'
                f'<div style="font-size:11px;color:#666">by {t.created_by}</div></div>'
            )
        done = [t for t in tm.tasks.values() if t.status == TaskStatus.COMPLETED]
        for t in done[-3:]:
            parts.append(
                f'<div style="opacity:.6;background:#f1f8e9;border-left:3px solid #81c784;'
                f'border-radius:8px;padding:6px;margin-bottom:4px;font-size:11px">'
                f'&#x2705; <b>{t.name}</b> &middot; by {t.created_by}</div>'
            )
        if not tm.current_task and not tm._queue and not tm.interrupted_stack:
            parts.append('<div style="text-align:center;color:#aaa;padding:24px;font-size:13px">&#x1F4A4; Idle</div>')
    return '<div style="font-family:-apple-system,sans-serif">' + "".join(parts) + "</div>"


def build_ui(nurse, tts, stt, startup_audio=None):
    tm = nurse.tm

    def get_log():
        return "\n".join(reversed(list(tm._log)[-18:])) if tm._log else "No activity yet."

    # ── Explicit named functions (Gradio 6 requires these, no lambdas) ──
    def send_doctor_text(text):
        if not text or not text.strip():
            return "", _board(tm), get_log(), None, "Type a command"
        resp  = nurse.process(text.strip(), "doctor")
        audio = tts.synth(resp)
        return "", _board(tm), get_log(), (tts.sr, audio), resp

    def send_patient_text(text):
        if not text or not text.strip():
            return "", _board(tm), get_log(), None, "Type a command"
        resp  = nurse.process(text.strip(), "patient")
        audio = tts.synth(resp)
        return "", _board(tm), get_log(), (tts.sr, audio), resp

    def send_doctor_voice(audio_data):
        if audio_data is None:
            return _board(tm), get_log(), None, "No audio"
        try:
            sr, data = audio_data
            if data.ndim > 1: data = data.mean(axis=1)
            data = data.astype(np.float32)
            if data.max() > 1.0: data /= 32768.0
            text = stt.transcribe(data, samplerate=sr)
            if not text:
                return _board(tm), get_log(), None, "Could not hear — speak louder"
            resp  = nurse.process(text, "doctor")
            audio = tts.synth(resp)
            return _board(tm), get_log(), (tts.sr, audio), f'Heard: "{text}"'
        except Exception as e:
            return _board(tm), get_log(), None, f"Error: {e}"

    def send_patient_voice(audio_data):
        if audio_data is None:
            return _board(tm), get_log(), None, "No audio"
        try:
            sr, data = audio_data
            if data.ndim > 1: data = data.mean(axis=1)
            data = data.astype(np.float32)
            if data.max() > 1.0: data /= 32768.0
            text = stt.transcribe(data, samplerate=sr)
            if not text:
                return _board(tm), get_log(), None, "Could not hear — speak louder"
            resp  = nurse.process(text, "patient")
            audio = tts.synth(resp)
            return _board(tm), get_log(), (tts.sr, audio), f'Heard: "{text}"'
        except Exception as e:
            return _board(tm), get_log(), None, f"Error: {e}"

    def refresh():
        pending = tts.pop_audio()
        return _board(tm), get_log(), pending

    # ── Quick action functions (one per button, no lambdas) ──
    def btn_bp():      return send_doctor_text("check blood pressure")
    def btn_med():     return send_doctor_text("administer medication")
    def btn_temp():    return send_doctor_text("take temperature")
    def btn_vitals():  return send_doctor_text("vital signs check")
    def btn_assess():  return send_doctor_text("patient assessment")
    def btn_status():  return send_doctor_text("task status")
    def btn_emerg():   return send_patient_text("emergency I need help now")

    with gr.Blocks(title="NurseBot") as demo:
        gr.HTML(
            f'<div style="text-align:center;padding:12px 0 8px;border-bottom:1px solid #e5e7eb;margin-bottom:12px">'
            f'<h1 style="font-size:22px;font-weight:700;margin:0">&#x1F3E5; NurseBot</h1>'
            f'<p style="color:#6b7280;font-size:12px;margin:4px 0 0">'
            f'Map: <a href="http://localhost:{FLASK_PORT}/" target="_blank" style="color:#3b82f6">'
            f'localhost:{FLASK_PORT}</a></p></div>'
        )

        if startup_audio is not None:
            gr.Audio(value=(tts.sr, startup_audio), autoplay=True,
                     visible=False, interactive=False)

        bg_audio = gr.Audio(autoplay=True, visible=False, interactive=False)

        with gr.Row(equal_height=False):

            # ── Left: map + task board ──
            with gr.Column(scale=5):
                gr.HTML(
                    f'<iframe src="http://localhost:{FLASK_PORT}/" width="100%" height="430"'
                    f' style="border:none;border-radius:12px;overflow:hidden;background:#1a2035"></iframe>'
                )
                gr.HTML('<p style="font-weight:600;font-size:14px;margin:10px 0 6px">Task queue</p>')
                board = gr.HTML(value=_board(tm))

            # ── Middle: Doctor panel ──
            with gr.Column(scale=3, min_width=260):
                gr.HTML('<p style="font-weight:700;font-size:15px;color:#3b82f6;margin:0 0 8px">🩺 Doctor</p>')
                doc_text   = gr.Textbox(placeholder="check blood pressure room 302 · urgent room 301", label="Text command", lines=2)
                doc_send   = gr.Button("Send", variant="primary")
                doc_voice  = gr.Audio(sources=["microphone"], type="numpy", label="Voice command", show_label=True)
                doc_status = gr.Textbox(label="Status", lines=2, interactive=False, value="Ready")
                doc_audio  = gr.Audio(label="Nurse response", autoplay=True, interactive=False)
                gr.HTML('<hr style="margin:10px 0">')
                gr.HTML('<p style="font-weight:600;font-size:13px;margin:0 0 6px">Quick actions</p>')
                with gr.Row():
                    qbp  = gr.Button("Check BP",    variant="secondary")
                    qmd  = gr.Button("Medication",  variant="secondary")
                with gr.Row():
                    qtp  = gr.Button("Temperature", variant="secondary")
                    qvt  = gr.Button("Vitals",      variant="secondary")
                with gr.Row():
                    qas  = gr.Button("Assessment",  variant="secondary")
                    qst  = gr.Button("Status",      variant="secondary")

            # ── Right: Patient panel ──
            with gr.Column(scale=3, min_width=260):
                gr.HTML('<p style="font-weight:700;font-size:15px;color:#10b981;margin:0 0 8px">🛏️ Patient</p>')
                pat_text   = gr.Textbox(placeholder="I'm in pain · I need help", label="Text command", lines=2)
                pat_send   = gr.Button("Send", variant="primary")
                pat_voice  = gr.Audio(sources=["microphone"], type="numpy", label="Voice command", show_label=True)
                pat_status = gr.Textbox(label="Status", lines=2, interactive=False, value="Ready")
                pat_audio  = gr.Audio(label="Nurse response", autoplay=True, interactive=False)
                gr.HTML('<hr style="margin:10px 0">')
                qhelp = gr.Button("🆘  EMERGENCY", variant="stop")
                gr.HTML('<hr style="margin:10px 0">')
                log_box = gr.Textbox(value="", label="Activity log", lines=10,
                                     interactive=False, max_lines=14)

        timer = gr.Timer(value=0.5)
        timer.tick(fn=refresh, outputs=[board, log_box, bg_audio])

        # ── Doctor wiring ──
        doc_outs = [doc_text, board, log_box, doc_audio, doc_status]
        doc_send.click(fn=send_doctor_text,  inputs=[doc_text],   outputs=doc_outs)
        doc_text.submit(fn=send_doctor_text, inputs=[doc_text],   outputs=doc_outs)
        doc_voice.change(fn=send_doctor_voice, inputs=[doc_voice],
                         outputs=[board, log_box, doc_audio, doc_status])

        qbp.click(fn=btn_bp,     outputs=doc_outs)
        qmd.click(fn=btn_med,    outputs=doc_outs)
        qtp.click(fn=btn_temp,   outputs=doc_outs)
        qvt.click(fn=btn_vitals, outputs=doc_outs)
        qas.click(fn=btn_assess, outputs=doc_outs)
        qst.click(fn=btn_status, outputs=doc_outs)

        # ── Patient wiring ──
        pat_outs = [pat_text, board, log_box, pat_audio, pat_status]
        pat_send.click(fn=send_patient_text,  inputs=[pat_text],   outputs=pat_outs)
        pat_text.submit(fn=send_patient_text, inputs=[pat_text],   outputs=pat_outs)
        pat_voice.change(fn=send_patient_voice, inputs=[pat_voice],
                         outputs=[board, log_box, pat_audio, pat_status])
        qhelp.click(fn=btn_emerg, outputs=pat_outs)

    return demo

print("✅ Gradio UI defined")
````
